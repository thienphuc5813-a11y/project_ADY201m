#  PHÂN TÍCH VÀ DỰ BÁO XU HƯỚNG BIẾN ĐỘNG GIÁ BẤT ĐỘNG SẢN RESIDENTIAL

## 1. Giới thiệu bài toán
Trong nền kinh tế hiện đại, bất động sản luôn là một trong những kênh tích sản và đầu tư quan trọng nhất nhưng cũng đầy rủi ro do tính chất phi tập trung và bất đối xứng thông tin của thị trường. Việc định giá một căn nhà từ trước đến nay thường phụ thuộc nhiều vào cảm tính hoặc các phương pháp so sánh thủ công của môi giới, vốn dễ dẫn đến sai lệch lớn.

Sự bùng nổ của Khoa học dữ liệu (Data Science) đã mở ra một chương mới. Bài toán dự đoán giá nhà tại Ames, Iowa không chỉ đơn thuần là một cuộc đua thuật toán trên Kaggle, mà chính là mô hình thu nhỏ của bài toán định giá tài sản trong thực tế. Nó đại diện cho xu hướng chuyển dịch từ "định giá dựa trên kinh nghiệm" sang "định giá dựa trên thuật toán tối ưu" (Algorithmic Valuation), giúp minh bạch hóa thị trường và giảm thiểu rủi ro cho các bên tham gia.

## 2. Mục tiêu phân tích
Dự án được thực hiện nhằm khai thác triệt để bộ dữ liệu gồm 79 biến mô tả chi tiết mọi khía cạnh của các bất động sản residential. Mục tiêu cốt lõi là kiến tạo một hệ thống dự báo và phân tích giúp các nhà đầu tư và phát triển bất động sản trả lời **3 câu hỏi kinh doanh chiến lược (Research Questions - RQs)**:

* **RQ1 (Feature Importance & Value Drivers):** Những nhóm yếu tố nào (Vị trí địa lý, Diện tích không gian, hay Chất lượng hoàn thiện/Vật liệu) đóng vai trò quyết định nhất đến giá trị của một bất động sản? Mức độ tác động biên (marginal effect) của chúng ra sao?

* **RQ2 (Sub-market Dynamics & Premium Factors):** Có sự khác biệt mang tính hệ thống về giá giữa các khu phố (Neighborhoods) không? Đâu là những khu vực "Rising Stars" có tốc độ tăng giá hoặc biên độ lợi nhuận cao nhất dựa trên năm xây dựng và sửa chữa?

* **RQ3 (Non-linear Relationships & Anomalies):** Làm thế nào để mô hình hóa các mối quan hệ phi tuyến phức tạp (ví dụ: việc tăng thêm 1 phòng ngủ có làm tăng giá đều đặn hay sẽ chạm ngưỡng bão hòa?) và nhận diện các bất động sản bị định giá sai (Mispriced Properties) để tìm kiếm cơ hội đầu tư kinh doanh chênh lệch giá (Arbitrage)?


Kết quả phân tích phục vụ trực tiếp cho quá trình ra quyết định của 3 nhóm đối tượng:
* **Nhà đầu tư & Quỹ định giá:** Tự động hóa quy trình thẩm định tài sản, phát hiện các bất động sản "dưới giá trị thực" để mua vào.

* **Đơn vị phát triển/Nhà thầu:** Biết được nên tập trung ngân sách vào đâu (ví dụ: nâng cấp nhà bếp hay mở rộng diện tích sân vườn) để tối đa hóa giá trị bán lại (ROI).

* **Người mua nhà cá nhân:** Tránh bẫy định giá cao từ các bên môi giới, đưa ra quyết định mua sắm phù hợp với ngân sách và giá trị thực tế.


## 3. Các yếu tố định hình giá trị bất động sản

Dựa trên bản chất của tập dữ liệu Ames, có 4 trục yếu tố trọng yếu trực tiếp chi phối và cấu thành nên giá trị cuối cùng của một căn nhà:

* **Vị trí và Khu vực (Location Metrics):** Biến `Neighborhood`, `MSZoning`, khoảng cách đến các trục đường chính (`Condition1`, `Condition2`). Đây là yếu tố cốt lõi quyết định tính thanh khoản và giá nền của bất động sản.

* **Quy mô và Không gian (Space & Capacity):** Tổng diện tích mặt sàn (`GrLivArea`), diện tích tầng hầm (`TotalBsmtSF`), số lượng phòng tắm (`FullBath`) và sức chứa của gara (`GarageCars`).

* **Chất lượng và Tình trạng (Quality & Condition):** Đánh giá tổng thể từ biến `OverallQual` và `OverallCond`, vật liệu mái, hệ thống sưởi, và năm trùng tu (`YearRemodAdd`).

* **Các yếu tố ngoại cảnh phụ trợ (Premium Features):** Sự hiện diện của bể bơi (`PoolArea`), hàng rào (`Fence`), hay lò sưởi (`Fireplaces`) – những yếu tố tạo ra giá trị thặng dư cao cho phân khúc hạng sang.


## 4. Cách tiếp cận phân tích

Để giải quyết bài toán hồi quy phức tạp này, dự án áp dụng quy trình End-to-End Data Science Pipeline nghiêm ngặt:

1.  **Tiền xử lý và Kỹ nghệ đặc trưng (Data Preprocessing & Feature Engineering):**
    * Xử lý dữ liệu khuyết thiếu lớn (Missing Data) bằng phương pháp gán giá trị thông minh (K-NN Imputation hoặc Mode/Median tùy đặc trưng).
    * Xử lý bất đối xứng và phân phối lệch (Skewed Data) của biến mục tiêu `SalePrice` bằng phép biến đổi Logarithm hoặc Box-Cox để đưa về phân phối chuẩn (Normal Distribution).
    * Mã hóa các biến phân loại (Categorical Encoding) bằng One-Hot Encoding hoặc Target Encoding.

2.  **Phân tích khai phá dữ liệu (EDA):** Sử dụng các biểu đồ Heatmap tương quan (Correlation Matrix) để lọc bỏ hiện tượng đa cộng tuyến (Multicollinearity) giữa các biến (ví dụ: mối tương quan quá cao giữa `GarageCars` và `GarageArea`).

3.  **Mô hình hóa nâng cao (Advanced Modeling):**
    * Triển khai các mô hình tuyến tính có phạt để chống quá khớp (Ridge, Lasso, ElasticNet).
    * Ứng dụng các thuật toán mạnh mẽ dựa trên cây quyết định (Gradient Boosting Machine): XGBoost, LightGBM, và CatBoost.

4.  **Tối ưu hóa và Ensembling:** Sử dụng Bayesian Optimization để tinh chỉnh siêu tham số (Hyperparameter tuning). Sau đó, kết hợp các mô hình thông qua kỹ thuật **Stacking** hoặc **Blending** để tận dụng tối đa thế mạnh của từng thuật toán.

5.  **Đánh giá mô hình (Evaluation Metrics):** Sử dụng chỉ số chuẩn hóa Root Mean Squared Error (RMSE) trên giá trị log của $SalePrice$ để đánh giá chính xác độ chệch và phương sai của mô hình.


$$RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (\log(y_i) - \log(\hat{y}_i))^2}$$


## 5. Mục tiêu kỳ vọng

* Xây dựng thành công hệ thống dự báo giá nhà đạt độ chính xác cao (nằm trong Top phân hạng của Kaggle leaderboard), tối thiểu hóa sai số giữa giá dự đoán và giá thị trường thực tế.

* Bóc tách và xếp hạng định lượng được **Top 10 biến số** có sức ảnh hưởng lớn nhất đến giá nhà, phá vỡ các định kiến mơ hồ trong định giá bất động sản truyền thống.

* Thiết lập một **Khung framework định giá tự động (Automated Valuation Model - AVM)** có khả năng mở rộng và tái sử dụng cho các bộ dữ liệu bất động sản ở các thị trường/đô thị khác trong tương lai.
